```python
structured_blocks = []

for page_index, page in enumerate(pdf):
    for block in page.get_text("blocks", sort=True):
        x0, y0, x1, y1, text, *_ = block

        if not text.strip():
            continue

        structured_blocks.append({
            "page": page_index + 1,
            "bbox": (x0, y0, x1, y1),
            "text": text.strip()
        })

print("Total blocks:", len(structured_blocks))
print("First block:", structured_blocks[0])
```

```python
import re

clean_blocks = []

for block in structured_blocks:
    text = block["text"].replace("\x00", "").strip()

    if re.fullmatch(
        r"NASA Systems Engineering Handbook\s+Rev\s+2\s+[ivxlcdm\d]+",
        text,
        flags=re.IGNORECASE
    ):
        continue

    if not text:
        continue

    clean_blocks.append({
        **block,
        "text": text
    })

print("Original blocks:", len(structured_blocks))
print("Clean blocks:", len(clean_blocks))
print("Removed:", len(structured_blocks) - len(clean_blocks))
```

```python
remaining_nulls = sum("\x00" in b["text"] for b in clean_blocks)

remaining_footers = sum(
    "NASA Systems Engineering Handbook" in b["text"]
    for b in clean_blocks
)

print("Remaining NULL blocks:", remaining_nulls)
print("Remaining footer-like blocks:", remaining_footers)
```

```python
import re
from collections import Counter

def classify_block(block):
    text = block["text"].strip()
    lines = text.splitlines()

    if re.match(r"^\d+(?:\.\d+)*\s+[A-Z]", text):
        return "heading_candidate"

    if re.match(r"^[•\uf0b7]", text):
        return "list_candidate"

    if len(lines) >= 3 and len(text) > 150:
        return "paragraph_candidate"

    return "other"

for block in clean_blocks:
    block["type"] = classify_block(block)

print(Counter(block["type"] for block in clean_blocks))
```

```python
import re

for block in clean_blocks:
    text = block["text"].replace("\xa0", " ").strip()

    if re.match(r"^Figure\s+\d+(?:\.\d+)*-\d+", text, re.I):
        print(
            f"Page {block['page']}:",
            text[:100],
            "BBox:",
            tuple(round(x, 1) for x in block["bbox"])
        )
```

```python
caption_candidates = []

for block in clean_blocks:
    text = block["text"].replace("\xa0", " ").strip()

    if not re.match(r"^Figure\s+\d+(?:\.\d+)*-\d+", text, re.I):
        continue

    if re.search(r"\.{5,}", text):
        continue

    if len(text) > 150:
        continue

    caption_candidates.append(block)

for block in caption_candidates:
    print(
        f"Page {block['page']}:",
        repr(block["text"][:120])
    )
```

```python
for caption in caption_candidates:
    page_num = caption["page"]
    caption_y = caption["bbox"][1]

    nearby = [
        block for block in clean_blocks
        if block["page"] == page_num
        and block["bbox"][3] <= caption_y
        and block["bbox"][3] >= caption_y - 120
    ]

    print(f"\n--- Page {page_num} ---")
    print("Caption:", caption["text"][:100])

    for block in nearby:
        print(
            tuple(round(v, 1) for v in block["bbox"]),
            repr(block["text"][:90])
        )
```
